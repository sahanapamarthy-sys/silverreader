```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
import numpy as np
```

```python

df = pd.read_csv("../../../course_assignments/homework8/state_data.csv")
print(df.shape) #there are 52 rows and 4 columns 
print(df.head())
print(df.tail())
#the first state is alaska and the last is wyoming 
#there are 52 total states and more than 50 including United States territories

print(df["latitude"].median())
print(df["longitude"].median())

df["absolute values (long)"] = abs((df["longitude"] - df["longitude"].median()))
df["absolute values (lat)"] = abs((df["latitude"] - df["latitude"].median()))
print(df.head(10))
print(df.shape) #there are 52 rows and 6 columns 

print(f"{df["absolute values (lat)"].min():.2f}")
print(f"{df["absolute values (lat)"].max():.2f}")

print(f"{df['absolute values (long)'].min():.2f}")
print(f"{df["absolute values (lat)"].max():.2f}")

print(df.info)
#gives info on the dataframe and confirms object is a panda dataframe , rangeindex (displays total number of rows), and index boundaries. 
```

```python
df = sns.load_dataset('iris')
df.dropna(inplace = True)
print(df.shape) #922 rows and 63 columns 
print(df.head(150))
```

```python
fig, axs = plt.subplots(nrows = 2, ncols = 2, figsize = (25,25))
axs[0,0].scatter(df["sepal_length"],df['sepal_width'])
axs[0,1].scatter(df['petal_length'],df['petal_width'])
sns.histplot(df['sepal_length'], ax = axs[1,0])
sns.boxplot(df['petal_length'], ax = axs[1,1])

axs[0,0].set_title("Sepal Width vs Sepal Length", fontsize = 25)
axs[0,0].set_xlabel("Sepal Length",fontsize = 20)
axs[0,0].set_ylabel("Sepal Width",fontsize = 20)

axs[0,1].set_title("Petal Width vs Petal Length", fontsize = 25)
axs[0,1].set_xlabel("Petal Length", fontsize = 20)
axs[0,1].set_ylabel("Petal Width", fontsize = 20)

axs[1,0].set_title("Sepal Length Distribution", fontsize=25)
axs[1,0].set_xlabel("Sepal Length", fontsize=20)
axs[1,0].set_ylabel("Count", fontsize=20)

axs[1,1].set_title("Petal Length Distribution", fontsize=25)
axs[1,1].set_xlabel("Petal Length", fontsize=20)
axs[1,1].set_ylabel("Petal Length", fontsize=20)


plt.show()
```
