```python
import numpy as np
import matplotlib.pyplot as plt 


data = np.loadtxt("../../../course_assignments/homework7/star_data.txt",skiprows=1)
print(data.shape)
#There are 2642 rows and 2 columns 

magnitude = data[:,0]
temperature = data[:,1]

print(magnitude.max()) #Zeta Crucis(the dimmest star)
print(magnitude.min()) #Schedar(the brightest star)

print(temperature.max())
print(temperature.min())
```

```python
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)

fig, axs = plt.subplots(figsize = (8,6))

axs.scatter(
    log_temperature,
    log_magnitude,
    c=temperature,
    cmap="coolwarm")

plt.gca().invert_xaxis()
plt.gca().invert_yaxis()

axs.scatter(
    log_temperature[log_magnitude.argmin()],
    log_magnitude[log_magnitude.argmin()], 
    marker = "*",
    color = "black",
    linewidth = 3,
    label = 'brightest star'
    
)
axs.scatter(
    log_temperature[log_magnitude.argmax()],
    log_magnitude[log_magnitude.argmax()],
    marker = "*",
    color = "purple",
    linewidth = 3,
    label = 'dimmest star'
)

axs.scatter(
    log_temperature[log_temperature.argmax()],
    log_magnitude[log_temperature.argmax()],
    marker = "*", 
    color = 'green',
    linewidth = 3,
    label = 'hotest star'
)

axs.scatter(
    log_temperature[log_temperature.argmin()],
    log_magnitude[log_temperature.argmin()],
    marker = "*", 
    color = 'orange',
    linewidth = 3,
    label = 'coldest star'
)

axs.set_xlabel("Temperature (log K)",fontsize=12)
axs.set_ylabel("Magnitude", fontsize = 12)
axs.set_title("Hertzsprung-Russel Diagram",fontsize=14, color = 'green')

plt.legend(fontsize = 'x-small',loc = 'upper left')
plt.show()
```
