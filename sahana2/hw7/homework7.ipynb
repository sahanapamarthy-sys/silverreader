```python
import numpy as np
import matplotlib.pyplot as plt 
```

```python
def make_sin_wave(x,A,w):
    return A * np.sin(w*x)
x = np.linspace(0,2*np.pi,1000)
amplitude = [0.5,1,1.5,2,2.5]
frequency = [1,2,3,4,5]

for A, w in zip(amplitude, frequency):
    y1 = make_sin_wave(x,A,1)
for A, w in zip(amplitude,frequency):
    y2 = make_sin_wave(x,0.5,w)
for A, w in zip(amplitude,frequency):
    y3 = make_sin_wave(x,A,w)

fig, axs = plt.subplots(figsize =(25,5), nrows = 1, ncols = 3, sharey = True)
axs[0].plot(x,y1,linewidth=4,linestyle=":")
axs[1].plot(x,y2,linewidth=4,linestyle="--")
axs[2].plot(x,y3,linewidth=4,linestyle="-")

axs[0].set_xlabel("x values",fontsize = 20)
axs[1].set_xlabel("x values",fontsize = 20)
axs[2].set_xlabel("x values",fontsize = 20)

axs[0].set_ylabel("y values",fontsize=20)
axs[1].set_ylabel("y values",fontsize=20)
axs[1].set_ylabel("y values", fontsize=20)

axs[0].set_title("Varying Amplitude",fontsize=25)
axs[1].set_title("Varying Frequency", fontsize = 25)
axs[2].set_title("Varying Both", fontsize = 25)

axs[0].tick_params(axis = 'both', labelsize = 15)
axs[1].tick_params(axis = 'both', labelsize = 15)
axs[2].tick_params(axis = 'both', labelsize = 15)

axs[0].grid(True)
axs[1].grid(True)
axs[2].grid(True)

plt.show()


```
