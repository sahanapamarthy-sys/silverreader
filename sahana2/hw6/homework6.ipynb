```python
list = [20,27,9,12,31,17,4,50,39,35]
arr = np.array([list])

print(arr*4)

print(np.mean(arr))

print(np.median(arr))

print(np.std(arr))

i=np.argmin(arr)
print(arr.flat[i])

j=np.argmax(arr)
print(arr.flat[j])

mask = arr > 20 
print(arr[mask])
```

```python
matrix = [
[5, 12, 17, 9],
[8, 21, 14, 3],
[6, 13, 19, 11]
]
matrix = np.array(matrix)

print(np.shape(matrix))
#there are 3 rows and 4 columns

print(matrix[:,2])
print(matrix*2)

print(np.mean(matrix,axis=1)) #row mean
print(np.mean(matrix,axis=0)) #column mean 
print(np.mean(matrix))

print(np.max(matrix,axis=1))
print(np.min(matrix,axis=0))

flatten = matrix.flatten()
print(np.shape(flatten)) 
#1d array with 12 elements. it went from 2D to 1D
print(np.argmin(flatten))
print(np.argmax(flatten))
print(np.unravel_index(7,matrix.shape))
print(np.unravel_index(5,matrix.shape))

mask = matrix > 15 
print(matrix[mask])

print(np.sort(matrix,axis=1))


```

```python


def mag_prime_num(arr):
    arr_new =[]
    for row in arr:
        for number in row:
            if number % 2 != 0:
                arr_new.append(row)
    return arr_new
    
arr = np.array([
[2, 3, 5],
[14, 6, 8],
[11, 13, 17],
[7, 10, 13]
])
print(mag_prime_num(arr))                
```
